# Creating NumPy Arrays

```python
# Step 1: Import NumPy
import numpy as np

# Step 2: Create a 1D array representing years of experience
years_exp = np.array([1, 3, 5, 7, 10])
print("Years of Experience:", years_exp)

# Step 3: Create a 2D array representing sample salaries (in thousands)
salaries = np.array([[50, 60, 70], [80, 90, 100]])
print("Salary Matrix:\n", salaries)

# Step 4: Create an array of zeros and ones for placeholder analysis
zeros_array = np.zeros((2, 2)) # 2x2 zeros
ones_array = np.ones((2, 3)) # 2x3 ones
print("Zeros Array:\n", zeros_array)
print("Ones Array:\n", ones_array)
```

```python
# --- Try It: Creating NumPy Arrays ---

# 1. Add 2 more sample data points to years_exp (1D array)
# We use np.append to add [12, 15] to the existing array
years_exp = np.append(years_exp, [12, 15])
print("Updated Years of Experience:", years_exp)

# 2. Add 2 more data points to salaries (2D array)
# Note: Since the original has 3 columns, our new data must also have 3 columns.
# axis=0 means we are adding a new row.
new_rows = np.array([[110, 120, 130], [140, 150, 160]])
salaries = np.append(salaries, new_rows, axis=0)

print("\nUpdated Salary Matrix:\n", salaries)

# 3. Create a 3x3 identity matrix
# The function np.eye(n) creates a square matrix with 1s on the diagonal.
identity_3x3 = np.eye(3)
print("\n3x3 Identity Matrix:\n", identity_3x3)
```

# Array Operations


```python
# Step 1: Element-wise addition
exp_plus_5 = years_exp + 5 # Add 5 years to all experience values
print("Years + 5:", exp_plus_5)

# Step 2: Element-wise multiplication
exp_times_2 = years_exp * 2 # Multiply all values by 2
print("Years * 2:", exp_times_2)

# Step 3: Dot product (simulate salary projections)
sample1 = np.array([1, 2, 3])
sample2 = np.array([4, 5, 6])
dot_result = np.dot(sample1, sample2)
print("Dot Product:", dot_result)
```

```python
# --- Try It: Array Operations ---

# 1. Experiment with subtraction and division
# 2. Apply np.exp() (Exponential)
# This calculates e raised to the power of each element
# 3. Apply np.log() (Natural Logarithm)
# This calculates the natural log (ln) of each element

minus_ones = years_exp - 1
halved = years_exp / 2
log_vals = np.log(years_exp)

print("Minus 1:", minus_ones)
print("Halved:", halved)
print("Logarithm:", log_vals)
```

# Indexing and Slicing

```python
# Step 1: Access individual element
print("First year of experience:", years_exp[0])

# Step 2: Slice arrays
print("First two salaries:", salaries[0, :2]) # First two elements of first row

# Step 3: Access all rows for a specific column
print("Second column salaries:", salaries[:, 1])

# Step 4: Negative indexing
print("Last year of experience:", years_exp[-1])
```

```python
# --- Try It: Slicing and Indexing ---

# 1. Reverse arrays using slicing
# The [::-1] tells Python: "Start at the end, end at the start, step back by 1"
reversed_years = years_exp[::-1]
print("Reversed Years of Experience:", reversed_years)

# 2. Slice 2D arrays for subgroups of salaries
# Let's extract a 2x2 "mini-grid" from the top-left of your salaries matrix
# salaries[rows, columns] -> [0:2 (rows 0 and 1), 0:2 (columns 0 and 1)]
salary_subgroup = salaries[0:2, 0:2]

print("\nOriginal Salary Matrix:\n", salaries)
print("\nSubgroup (Top-left 2x2):\n", salary_subgroup)
```

# Reshaping and Dimensions

```python
# Step 1: Reshape 1D array into 2x3 matrix for batch analysis
reshaped_exp = np.reshape(np.arange(1, 7), (2, 3))
print("Reshaped Experience Array:\n", reshaped_exp)

# Step 2: Flatten 2D arrays back into 1D
flattened_exp = reshaped_exp.flatten()
print("Flattened Array:", flattened_exp)

# Step 3: Transpose example
print("Transposed Array:\n", reshaped_exp.T)
```

```python
# --- Try It: Reshaping and Dimensions ---

# 1. Create a larger 1D array to experiment with (12 elements)
# np.arange(12) gives numbers 0 through 11
data_12 = np.arange(12)
print("Original 1D Data:", data_12)

# 2. Reshape into different dimensions
# Note: The new dimensions MUST multiply to the total number of elements (12)

# Option A: Reshape into 3 rows and 4 columns (3x4)
reshaped_3x4 = data_12.reshape(3, 4)
print("\nReshaped to 3x4 Matrix:\n", reshaped_3x4)

# Option B: Reshape into 4 rows and 3 columns (4x3)
reshaped_4x3 = data_12.reshape(4, 3)
print("\nReshaped to 4x3 Matrix:\n", reshaped_4x3)

# Option C: 3D Array (2 blocks, 3 rows, 2 columns) - Used in image processing
reshaped_3D = data_12.reshape(2, 3, 2)
print("\nReshaped to 3D Array (2x3x2):\n", reshaped_3D)

# 3. Quick Experiment with Transpose (.T)
# This flips the rows and columns
print("\nTransposed 4x3 to 3x4:\n", reshaped_4x3.T)
```

# Broadcasting in NumPy

```python
# Step 1: Add a bonus array to salaries using broadcasting
bonus = np.array([5, 10, 15])
salaries_with_bonus = salaries + bonus # Adds corresponding bonus to each column
print("Salaries after bonus:\n", salaries_with_bonus)
```

```python
# --- Try It: Broadcasting ---

# 1. Multiply by a single scaling factor (Scalar Broadcasting)
# Let's say everyone gets a 10% raise (multiply by 1.1)
scaling_factor = 1.1
scaled_salaries = salaries * scaling_factor

print("Original Salaries:\n", salaries)
print("\nSalaries after 10% raise (Scalar Scaling):\n", scaled_salaries)

# 2. Multiply by a specific scaling factor per column
# Imagine different departments get different performance multipliers
# Column 1: x1.0, Column 2: x1.2, Column 3: x1.5
performance_multipliers = np.array([1.0, 1.2, 1.5])
weighted_salaries = salaries * performance_multipliers

print("\nSalaries after Performance Multipliers:\n", weighted_salaries)
```

# Statistical Operations


```python
# Step 1: Mean of years of experience
print("Mean experience:", np.mean(years_exp))

# Step 2: Standard deviation of experience
print("Std deviation of experience:", np.std(years_exp))

# Step 3: Max and Min salaries
print("Max salary:", np.max(salaries), "Min salary:", np.min(salaries))

# Step 4: Sum of salaries
print("Sum of all salaries:", np.sum(salaries))
```

```python
# --- Try It: Statistical Operations ---

# 1. Compute the Median
# Note: Unlike the Mean (average), the Median is not affected by outliers.
median_exp = np.median(years_exp)
print("Median experience:", median_exp)

# 2. Compute Percentiles
# Let's find the 25th (1st quartile) and 75th (3rd quartile) percentiles
p25 = np.percentile(years_exp, 25)
p75 = np.percentile(years_exp, 75)

print(f"25th Percentile: {p25}")
print(f"75th Percentile: {p75}")

# 3. Compute Median for 2D arrays (Salaries)
# You can do this for the whole matrix, or per column (axis=0)
median_salaries_per_col = np.median(salaries, axis=0)
print("\nMedian salary per column (department/role):", median_salaries_per_col)
```

# NumPy Functions on Arrays


```python
# Step 1: Apply trigonometric functions
angles = np.array([0, np.pi/4, np.pi/2])
print("Sine of angles:", np.sin(angles))
print("Cosine of angles:", np.cos(angles))

# Step 2: Apply function along rows (sum salaries per person)
salary_sums = np.apply_along_axis(np.sum, 1, salaries)
print("Sum of Salaries per person:", salary_sums)
```

```python
# --- Try It: Math Transformations & Custom Functions ---

# 1. Use np.sqrt() and np.log() for transformations
# Square root is often used in signal processing (like finding RMS)
sqrt_years = np.sqrt(years_exp)
# Log transformation is great for "squashing" data that has huge outliers
log_salaries = np.log(salaries)

print("Square root of years:", sqrt_years)
print("\nLog-transformed salaries:\n", log_salaries)

# 2. Apply a CUSTOM function using np.apply_along_axis()
# Let's define a function that calculates a "Performance Score"
# (Average salary * 0.1 for simplicity)
def calculate_bonus_rate(row):
    return np.mean(row) * 0.1

# We apply it across axis 1 (rows)
bonus_rates = np.apply_along_axis(calculate_bonus_rate, 1, salaries)

print("\nCustom Bonus Rate per person (row):", bonus_rates)
```

# Integrating NumPy and Pandas for Analysis


```python
import pandas as pd

# Step 1: Create a NumPy array of random data
data = np.random.randint(1, 50, size=(5, 3))

# Step 2: Convert the array into a DataFrame
df_data = pd.DataFrame(data, columns=['X', 'Y', 'Z'])
print("Generated Data:\n", df_data)

# Step 3: Apply NumPy functions to DataFrame columns
df_data['Log_X'] = np.log(df_data['X'])
df_data['Sqrt_Y'] = np.sqrt(df_data['Y'])
print("DataFrame with NumPy Transformations:\n", df_data)

# Step 4: Analyze data using Pandas correlation
print("Correlation Matrix:\n", df_data.corr())
```

```python
# --- Try It: NumPy + Pandas Integration ---

# 1. Compute overall column statistics using np.mean() and np.median()
# You can apply NumPy functions directly to Pandas columns (Series)
mean_stats = np.mean(df_data[['X', 'Y', 'Z']], axis=0)
median_stats = np.median(df_data[['X', 'Y', 'Z']], axis=0)

print("Column Means (X, Y, Z):\n", mean_stats)
print("\nColumn Medians (X, Y, Z):\n", median_stats)

# 2. Apply np.square() and np.exp()
# We'll create new columns for these results
df_data['Square_Z'] = np.square(df_data['Z'])
df_data['Exp_X_Small'] = np.exp(df_data['X'] / 10)  # Scaling X down so exp doesn't get too huge

print("\nDataFrame with Square and Exp Transformations:\n", df_data)
```

# Importing and Summarizing Data Using Pandas

```python
# Step 1: Save the DataFrame to a CSV file
df_data.to_csv('sample_data.csv', index=False)
print("Data saved to 'sample_data.csv'.")

# Step 2: Load the CSV file back into a new DataFrame
df_imported = pd.read_csv('sample_data.csv')
print("Imported DataFrame:\n", df_imported)

# Step 3: Generate summary statistics
summary_stats = df_imported.describe()
print("Summary Statistics:\n", summary_stats)

# Step 4: Display column means and standard deviations
print("Column Means:\n", df_imported.mean())
print("Column Standard Deviations:\n", df_imported.std())
```

# Data Grouping and Aggregation

```python
import pandas as pd
import numpy as np

# Step 1: Load the dataset
df_kaggle = pd.read_csv('survey_results_public.csv')
print("Loaded Dataset head:\n", df_kaggle.head())

# Step 2: Select relevant columns (UPDATED NAMES)
# Changed 'ConvertedComp' to 'ConvertedCompYearly' for the 2023 dataset
df_subset = df_kaggle[['Country', 'EdLevel', 'YearsCodePro', 'ConvertedCompYearly']]

# Step 3: Clean data by dropping missing values
df_clean = df_subset.dropna().copy()

# Step 4: Categorize experience (Data cleaning included)
# YearsCodePro often contains strings like "More than 50 years", so we convert to numbers
df_clean['YearsCodePro'] = pd.to_numeric(df_clean['YearsCodePro'], errors='coerce')
df_clean = df_clean.dropna() # Drop rows that couldn't be turned into numbers

df_clean['ExperienceLevel'] = np.where(df_clean['YearsCodePro'] >= 10, 'Senior', 'Junior')

# Step 5: Group by Country and ExperienceLevel
grouped_data = df_clean.groupby(['Country', 'ExperienceLevel'])['ConvertedCompYearly'].mean()

# Step 6: Reset index for readability
grouped_data = grouped_data.reset_index()
print("\nFinal Grouped Data:\n", grouped_data.head(10))
```

```python
# --- Try It: Final Analysis Tasks ---

# 1. Change the grouping to EdLevel and compute Median/Std
# This helps see how salary correlates with education
ed_stats = df_clean.groupby('EdLevel')['ConvertedCompYearly'].agg(['median', 'std'])
print("Salary Statistics by Education Level:\n", ed_stats)

# 2. Explore top 10 countries with the highest average compensation
# We group by Country, find the mean, and sort descending
country_ranking = df_clean.groupby('Country')['ConvertedCompYearly'].mean().sort_values(ascending=False)

print("\nTop 10 Countries by Average Compensation:\n", country_ranking.head(10))
```
